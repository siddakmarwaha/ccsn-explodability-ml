```python
# import modules
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# preprocessing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# LinearClassifaction
from sklearn.linear_model import LogisticRegression

# DecisionTreeClassifier
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier

# RandomForestClassifer
from sklearn.ensemble import RandomForestClassifier

# SVM - Classifaction
from sklearn.svm import SVC

# Hyperparameter tuning
from sklearn.model_selection import GridSearchCV
```

```python
# read in the dataset
data = pd.read_csv("../../data/results_s18.csv")

# split data into train/test
X = data.drop(columns = ['exploded'])
y = data.exploded

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = .25, random_state=304)
```

```python
# normalization
X_train_scale = StandardScaler().fit(X_train)
X_train_norm = X_train_scale

X_test_scale = StandardScaler().fit(X_test)
X_test_norm = X_test_scale.transform(X_test)
```

```python
pd.DataFrame(X_train_scale).describe()
```

```python
model = LogisticRegression().fit(X_train_norm, y_train)
```

```python
model.score(X_test_norm, y_test)
```

```python
pen = [None, 'l2']
C = [10**i for i in range(1,6)]
fit_params = {'penalty': pen, 'C': C}
cv_model = GridSearchCV(LogisticRegression(max_iter = 1000), fit_params, cv=10, scoring = 'accuracy', n_jobs = -1)
cv_model.fit(X_train_norm, y_train)
best_model = cv_model.best_estimator_
```

```python
best_model.score(X_test_norm, y_test)
```

```python
best_model.coef_, best_model.intercept_
```

```python
best_model.coef_ / X_train_scale.scale_
```

```python
X_train_scale.inverse_transform(best_model.coef_)
```
